## 1) Create the course environment from the repository root

Windows PowerShell:
```powershell
powershell -ExecutionPolicy Bypass -File .\scripts\setup.ps1
.\.venv\Scripts\python.exe scripts\course.py doctor
```
macOS/Linux:
```bash
bash scripts/setup.sh
.venv/bin/python scripts/course.py doctor
```

Activation is optional. In VS Code, select the interpreter inside `.venv` as both the Python interpreter and notebook kernel.

## 2) Python REPL basics
A REPL is an interactive Python prompt. Start the interpreter that belongs to this repository:

Windows PowerShell:
```powershell
.\.venv\Scripts\python.exe
```
macOS/Linux:
```bash
.venv/bin/python
```
At the `>>>` prompt, try the expressions below. Enter `exit()` when you are finished.

````python
1 + 1
'Hello ' + 'world'
len('data science')
````

## 3) Your first script
Create a file named `hello.py` with:

Run it with the repository interpreter:

Windows PowerShell:
```powershell

```
macOS/Linux:
```bash
.venv/bin/python hello.py
```

Below we reproduce similar logic in this notebook cell.

````python
def greet(name: str) -> str:
    """Return a friendly greeting for ``name``."""

    return f"Hello, {name}!"


greet("Data Scientist")
````

## 4) Inspecting packages in the course environment
The setup script already installed NumPy and pandas. Verify them through the same interpreter instead of using a bare `pip` command:

Windows PowerShell:
```powershell
.\.venv\Scripts\python.exe -m pip show numpy pandas
```
macOS/Linux:
```bash
.venv/bin/python -m pip show numpy pandas
```
Maintainers add course dependencies to `pyproject.toml`; learners normally rerun the setup script rather than installing ad hoc packages.

````python
import platform
import sys

import numpy as np
import pandas as pd

print(sys.executable)
print(sys.version)
print(platform.platform())
print(pd.__version__)
np.array([1, 2, 3])
````
