```python
def percent(n: float, total: float) -> str:
    if total == 0:
        return '0.0%'
    pct = (n / total) * 100
    return f"{pct:.1f}%"

assert percent(3, 8) == '37.5%'
percent(5, 9)
```

```python
def to_int(s: str):
    try:
        return int(s)
    except ValueError:
        return None

raw_a, raw_b = '12', '7'
A, B = to_int(raw_a), to_int(raw_b)
(A + B if A is not None and B is not None else None, A * B if A is not None and B is not None else None)
```
