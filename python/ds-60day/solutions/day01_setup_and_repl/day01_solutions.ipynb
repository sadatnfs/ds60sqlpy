```python
# gcd: Euclid's algorithm with simple assertions
def gcd(a: int, b: int) -> int:
    while b:
        a, b = b, a % b
    return a

assert gcd(12, 18) == 6
assert gcd(17, 13) == 1
print('gcd tests passed')
```
