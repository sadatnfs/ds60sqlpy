# Day 31 — Probability refresh
Objectives:
- Distributions (Bernoulli, Binomial, Normal).
- Law of large numbers via simulation.
- Bayes basics (intuitive).


```python
import numpy as np, matplotlib.pyplot as plt
rng = np.random.default_rng(42)
# Bernoulli trials
p=0.6
x = rng.binomial(n=1, p=p, size=10_000)
x.mean()  # ~0.6
# LLN
cum_mean = np.cumsum(x) / (np.arange(len(x))+1)
plt.plot(cum_mean); plt.axhline(p,color='r'); plt.title('Convergence to p'); plt.show()
```
