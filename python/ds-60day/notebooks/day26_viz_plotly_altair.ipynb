# Day 26 — Interactive visualization: Plotly or Altair
Objectives:
- Build interactive charts.
- Add tooltips, selections, and simple dashboards.
- Export HTML.

```python
from pathlib import Path

import plotly.express as px
import seaborn as sns

artifact_dir = Path('artifacts/day26')
artifact_dir.mkdir(parents=True, exist_ok=True)
df = sns.load_dataset('tips')
fig = px.scatter(df, x='total_bill', y='tip', color='time', hover_data=['day','size'])
fig.show()
fig.write_html(artifact_dir / 'interactive_scatter.html', include_plotlyjs=True)
```
