## Reproducibility instructions
- Create a virtual environment with the documented OS command.
- Install the project from `pyproject.toml` and its lock.
- Run training, evaluation, and plots from explicit modules.
- (Optional) launch the API container from its Dockerfile.

Windows PowerShell:
```powershell
py -3.12 -m venv .venv
.\.venv\Scripts\python.exe -m pip install -e .
.\.venv\Scripts\python.exe -m src.data.load
.\.venv\Scripts\python.exe -m src.model.train
.\.venv\Scripts\python.exe -m src.model.eval
```
macOS/Linux:
```bash
python3.12 -m venv .venv
.venv/bin/python -m pip install -e .
.venv/bin/python -m src.data.load
.venv/bin/python -m src.model.train
.venv/bin/python -m src.model.eval
```
