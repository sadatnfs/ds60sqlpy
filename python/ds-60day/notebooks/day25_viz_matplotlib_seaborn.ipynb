# Day 25 — Visualization with Matplotlib & seaborn
Objectives:
- Core plot types: hist, box, violin, bar, line, scatter.
- Aesthetics and themes.
- Save figures for reports.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

artifact_dir = Path('artifacts/day25')
artifact_dir.mkdir(parents=True, exist_ok=True)
df = sns.load_dataset('tips')
sns.histplot(data=df, x='total_bill', kde=True)
plt.show()
sns.boxplot(data=df, x='day', y='tip')
plt.show()
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')
plt.savefig(artifact_dir / 'fig_scatter.png', dpi=150)
```
